```python
from model.linear_regression_direct.linear_regression_class import linear_regression
import numpy as np
```

```python
rng = np.random.default_rng(30)

w_true = 3.5
b_true = -10.0

input_features = rng.standard_normal(100) * 10
target_values = w_true * input_features + b_true
```

```python
# create the linear regression class
numerical_linear_regression_model = linear_regression(input_features, target_values)
expression_linear_regression_model = linear_regression(input_features, target_values)
```

```python
zero_gradient_linear_regression_model = linear_regression(input_features, target_values)
zero_difference_linear_regression_model = linear_regression(input_features, target_values)
```

```python
# generate the model for numerical against expression
expression_linear_regression_model.gradient_descent(numerically=False)
numerical_linear_regression_model.gradient_descent(numerically=True)
```

```python
# generate the model for convergence with zero gradient agains zero difference between cost error functions
zero_gradient_linear_regression_model.gradient_descent(numerically=False, convergence_method='zero_gradient')
zero_difference_linear_regression_model.gradient_descent(numerically=False, convergence_method='zero_difference')
```

```python
# show the results
numerical_linear_regression_model.plots.plot_model_and_training_set()
numerical_linear_regression_model.prints.all()
```

```python
numerical_linear_regression_model.plots.plot_parameters_path(elev=0, azim=40)
expression_linear_regression_model.plots.plot_parameters_path(elev=0, azim=50)
numerical_linear_regression_model.plots.plot_error_squares()
```

```python
print("numerical derivative of J")
print("------------------------------------")

numerical_linear_regression_model.prints.all()

print("\n")

print("derivative expression of J")
print("------------------------------------")

expression_linear_regression_model.prints.all()

print("\n")

j_difference = abs(
    (numerical_linear_regression_model.j
     - expression_linear_regression_model.j)
    / expression_linear_regression_model.j
) * 100

j_abs_difference = abs(numerical_linear_regression_model.j - expression_linear_regression_model.j)

w_difference = abs(
    (numerical_linear_regression_model.w
     - expression_linear_regression_model.w)
    / expression_linear_regression_model.w
) * 100

b_difference = abs(
    (numerical_linear_regression_model.b
     - expression_linear_regression_model.b)
    / expression_linear_regression_model.b
) * 100

print("comparison")
print("------------------------------------")
print(f"J difference: {j_difference} %")
print(f"absolute J difference: {j_abs_difference}")
print(f"w difference: {w_difference} %")
print(f"b difference: {b_difference} %")
```

```python
j_difference = abs(
    (zero_gradient_linear_regression_model.j
     - zero_difference_linear_regression_model.j)
    / zero_difference_linear_regression_model.j
) * 100

j_abs_difference = abs(zero_difference_linear_regression_model.j - zero_gradient_linear_regression_model.j)

w_difference = abs(
    (zero_gradient_linear_regression_model.w
     - zero_difference_linear_regression_model.w)
    / zero_difference_linear_regression_model.w
) * 100

b_difference = abs(
    (zero_gradient_linear_regression_model.b
     - zero_difference_linear_regression_model.b)
    / zero_difference_linear_regression_model.b
) * 100

print("convergence with zero gradient")
print("------------------------------------")
zero_gradient_linear_regression_model.prints.all()

print("\nconvergence with zero difference")
print("------------------------------------")
zero_difference_linear_regression_model.prints.all()

print("\ncomparison")
print("------------------------------------")
print(f"J difference: {j_difference} %")
print(f"absolute J difference: {j_abs_difference}")
print(f"w difference: {w_difference} %")
print(f"b difference: {b_difference} %")
```
